```python
import re
import math
import numpy as np
import nltk

from collections import Counter
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
import urllib.request
import os

url = "https://raw.githubusercontent.com/pytorch/examples/main/word_language_model/data/wikitext-2/train.txt"

dataset_path = "wikitext-2-train.txt"

if not os.path.exists(dataset_path):
    urllib.request.urlretrieve(url, dataset_path)

print("Dataset downloaded successfully!")
print("File:", dataset_path)
```

```python
with open(dataset_path, "r", encoding="utf-8") as file:
    text = file.read()

print("Total characters:", len(text))
print("\nFirst 1000 characters:\n")
print(text[:1000])
```

```python
def clean_text(text):
    text = text.lower()
    
    # Replace special characters with spaces
    text = re.sub(r"[^a-z0-9'\s]", " ", text)
    
    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text)
    
    return text.strip()


cleaned_text = clean_text(text)

print("Cleaned text sample:\n")
print(cleaned_text[:1000])
```

```python
tokens = nltk.word_tokenize(cleaned_text)

print("Total tokens:", len(tokens))
print("\nFirst 50 tokens:")
print(tokens[:50])
```

```python
unigram_counts = Counter(tokens)

print("Total unique words:", len(unigram_counts))

print("\nMost common 20 words:")
print(unigram_counts.most_common(20))
```

```python
bigrams = list(zip(tokens[:-1], tokens[1:]))

bigram_counts = Counter(bigrams)

print("Total unique bigrams:", len(bigram_counts))

print("\nMost common 20 bigrams:")
print(bigram_counts.most_common(20))
```

```python
trigrams = list(zip(
    tokens[:-2],
    tokens[1:-1],
    tokens[2:]
))

trigram_counts = Counter(trigrams)

print("Total unique trigrams:", len(trigram_counts))

print("\nMost common 20 trigrams:")
print(trigram_counts.most_common(20))
```

```python
total_words = len(tokens)

unigram_probabilities = {
    word: count / total_words
    for word, count in unigram_counts.items()
}

print("Example unigram probabilities:\n")

for word, probability in list(unigram_probabilities.items())[:20]:
    print(f"{word}: {probability:.6f}")
```

```python
bigram_probabilities = {}

for (word1, word2), count in bigram_counts.items():
    probability = count / unigram_counts[word1]
    bigram_probabilities[(word1, word2)] = probability

print("Example bigram probabilities:\n")

for pair, probability in list(bigram_probabilities.items())[:20]:
    print(f"{pair}: {probability:.6f}")
```

```python
trigram_probabilities = {}

for (word1, word2, word3), count in trigram_counts.items():
    
    previous_bigram_count = bigram_counts[(word1, word2)]
    
    probability = count / previous_bigram_count
    
    trigram_probabilities[
        (word1, word2, word3)
    ] = probability

print("Example trigram probabilities:\n")

for triple, probability in list(trigram_probabilities.items())[:20]:
    print(f"{triple}: {probability:.6f}")
```

```python
def get_bigram_candidates(previous_word):
    
    candidates = []
    
    for (word1, word2), probability in bigram_probabilities.items():
        
        if word1 == previous_word:
            candidates.append((word2, probability))
    
    return candidates
```

```python
word = "machine"

candidates = get_bigram_candidates(word)

print("Candidates for:", word)
print(candidates[:20])
```

```python
def get_trigram_candidates(word1, word2):
    
    candidates = []
    
    for (w1, w2, w3), probability in trigram_probabilities.items():
        
        if w1 == word1 and w2 == word2:
            candidates.append((w3, probability))
    
    return candidates
```

```python
word1 = "machine"
word2 = "learning"

candidates = get_trigram_candidates(word1, word2)

print("Candidates:")
print(candidates[:20])
```

```python
def predict_next_words(sentence, top_n=5):
    
    # Clean input
    cleaned = clean_text(sentence)
    
    # Tokenize
    words = nltk.word_tokenize(cleaned)
    
    if len(words) == 0:
        return []
    
    predictions = []
    
    # ------------------------------------------------
    # CASE 1: Trigram prediction
    # ------------------------------------------------
    
    if len(words) >= 2:
        
        word1 = words[-2]
        word2 = words[-1]
        
        trigram_candidates = get_trigram_candidates(
            word1,
            word2
        )
        
        if trigram_candidates:
            
            predictions = trigram_candidates
            
            predictions.sort(
                key=lambda x: x[1],
                reverse=True
            )
            
            return predictions[:top_n]
    
    # ------------------------------------------------
    # CASE 2: Bigram prediction
    # ------------------------------------------------
    
    last_word = words[-1]
    
    bigram_candidates = get_bigram_candidates(last_word)
    
    if bigram_candidates:
        
        predictions = bigram_candidates
        
        predictions.sort(
            key=lambda x: x[1],
            reverse=True
        )
        
        return predictions[:top_n]
    
    # ------------------------------------------------
    # CASE 3: Unigram fallback
    # ------------------------------------------------
    
    predictions = [
        (word, probability)
        for word, probability
        in unigram_probabilities.items()
    ]
    
    predictions.sort(
        key=lambda x: x[1],
        reverse=True
    )
    
    return predictions[:top_n]
```

```python
sentence = "machine learning"

predictions = predict_next_words(
    sentence,
    top_n=5
)

print("Input:", sentence)

print("\nTop predicted next words:")

for rank, (word, probability) in enumerate(predictions, 1):
    print(
        f"{rank}. {word} - {probability:.4f}"
    )
```

```python
def display_predictions(sentence, top_n=5):
    
    predictions = predict_next_words(
        sentence,
        top_n
    )
    
    print("\n" + "=" * 50)
    print("SMART NEXT-WORD PREDICTOR")
    print("=" * 50)
    
    print("Input:", sentence)
    
    if not predictions:
        print("\nNo predictions found.")
        return
    
    print("\nTop predicted next words:")
    
    for rank, (word, probability) in enumerate(
        predictions,
        1
    ):
        print(
            f"{rank}. {word} - {probability:.4f}"
        )
    
    print("=" * 50)
```

```python
display_predictions(
    "machine learning",
    top_n=5
)
```

```python
test_sentences = [
    "machine learning",
    "the united",
    "new york",
    "artificial intelligence",
    "in the",
    "the first",
    "one of",
    "according to"
]

for sentence in test_sentences:
    
    display_predictions(
        sentence,
        top_n=5
    )
```

```python
while True:
    
    user_input = input(
        "\nEnter a sentence "
        "(type 'exit' to stop): "
    )
    
    if user_input.lower() == "exit":
        print("Predictor stopped.")
        break
    
    try:
        n = int(
            input("How many predictions? (3-5): ")
        )
        
        if n < 3 or n > 5:
            print("Please enter a value between 3 and 5.")
            continue
            
    except ValueError:
        print("Invalid number.")
        continue
    
    display_predictions(
        user_input,
        top_n=n
    )
```
